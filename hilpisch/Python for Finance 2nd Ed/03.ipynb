```python
# Type Hinting


# Built-in collection generics (list[float]) and Union operator (|)
def calculate_return(start_price: float, end_price: float | None = None) -> float:
    if end_price is None:
        return 0.0
    return (end_price - start_price) / start_price
```

```python
# Structured Data: dataclass or NamedTuple

from dataclasses import dataclass
from typing import NamedTuple


# Using dataclass
@dataclass(slots=True, frozen=True)
class OptionQuote:
    ticker: str
    strike: float
    expiry: str
    is_call: bool


# Using NamedTuple
class StockNT(NamedTuple):
    symbol: str
    shares: int
```

```python
# Structural Pattern Matching


def handle_option(option):
    match option:
        case OptionQuote(ticker=t, strike=s, expiry=e, is_call=True):
            print(f"Call option for {t} at strike {s} expiring on {e}")
        case OptionQuote(ticker=t, strike=s, expiry=e, is_call=False):
            print(f"Put option for {t} at strike {s} expiring on {e}")
        case _:
            print("Unknown option type")


def process_order(order: dict):
    match order:
        case {"action": "BUY", "ticker": str(t), "qty": int(q)}:
            print(f"Buying {q} shares of {t}")
        case {"action": "SELL", "ticker": str(t), "qty": int(q)}:
            print(f"Selling {q} shares of {t}")
        case _:
            raise ValueError("Invalid order format")
```

```python
from typing import Any


# 1. Define the function with modern type hints (PEP 604)
def process_order(order: dict[str, Any]) -> None:
    match order:
        case {"action": "BUY", "ticker": str(t), "qty": int(q)}:
            print(f"Executing BUY: {q} shares of {t}")
        case {"action": "SELL", "ticker": str(t), "qty": int(q)}:
            print(f"Executing SELL: {q} shares of {t}")
        case _:
            raise ValueError(f"Invalid or unrecognized order payload: {order}")


# --- Calling the function with valid orders ---

buy_order = {"action": "BUY", "ticker": "AAPL", "qty": 100}
process_order(buy_order)
# Output: Executing BUY: 100 shares of AAPL

sell_order = {"action": "SELL", "ticker": "NVDA", "qty": 50}
process_order(sell_order)
# Output: Executing SELL: 50 shares of NVDA

# --- Calling the function with an invalid order ---
invalid_order = {"action": "HOLD", "ticker": "TSLA", "qty": 10}
try:
    process_order(invalid_order)
except ValueError as e:
    print(e)
# Output: Invalid or unrecognized order payload: {'action': 'HOLD', 'ticker': 'TSLA', 'qty': 10}
```
