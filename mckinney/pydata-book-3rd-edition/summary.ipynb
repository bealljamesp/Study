Chapter 1: Preliminaries

Core Focus: Introduction to the ecosystem (Python, IPython, Jupyter, NumPy, pandas, matplotlib).

Key Takeaway: Explains the "two-language problem" (prototyping in dynamic languages vs. production in C/C++) and establishes why Python acts as the premier glue language for data-intensive processing.

Chapter 2: Python Language Basics, IPython, and Jupyter Notebooks

Core Focus: Essential syntax refresh, code semantics, interactive shell mechanics, and magic commands (%run, %timeit).

Key Takeaway: Establishes the baseline interactive computing environment required for exploratory data analysis.

Chapter 3: Built-In Data Structures, Functions, and Files

Core Focus: Native Python containers (tuples, lists, dicts, sets), list comprehensions, anonymous functions (lambda), and file handle processing.

Key Takeaway: Emphasizes native memory structures before transitioning to high-performance C-contiguous arrays.

Chapter 4: Numerical Computing Basics: NumPy

Core Focus: The ndarray, vectorization, broadcasting, universal functions (ufuncs), indexing, slicing, and conditional logic via boolean arrays.

Key Takeaway: Lays down the contiguous memory layout and SIMD vectorization mechanics necessary to bypass explicit Python loops.

Chapter 5: Getting Started with pandas

Core Focus: Introduction to Series, DataFrame, and Index objects. Essential indexing, reindexing, dropping entries, and arithmetic operations with alignment.

Key Takeaway: Data alignment by index labels eliminates structural mismatch errors common in raw array programming.

Chapter 6: Data Loading, Storage, and File Formats

Core Focus: CSV, text file parsing, JSON, XML/HTML web scraping, Microsoft Excel binary workbooks, and relational database interactions via SQL.

Key Takeaway: Real-world data ingestion requires handling messy encodings, delimited variations, and chunked iterations for large files.

Chapter 7: Data Cleaning and Preparation

Core Focus: Handling missing data (NaN), data transformation, string manipulation, regular expressions, and categorical data types for memory optimization.

Key Takeaway: Data wrangling consumes the majority of a quant's pipeline lifecycle; clean representations dictate downstream algorithmic stability.

Chapter 8: Data Wrangling: Join, Combine, and Reshape

Core Focus: Hierarchical indexing, database-style DataFrame joins (merge), concatenation, pivoting, and melting.

Key Takeaway: Reshaping transforms wide and long tables to match exact mathematical inputs required for linear algebra matrices.

Chapter 9: Plotting and Visualization

Core Focus: matplotlib integration directly within pandas, handling figures, subplots, ticks, labels, legends, and basic statistical plots (histograms, scatter, bar).

Key Takeaway: Visual inspection remains a mandatory validation step prior to model deployment.

Chapter 10: Data Aggregation and Group Operations

Core Focus: The groupby engine, split-apply-combine mechanics, grouped transformations, aggregations, and pivot tables.

Key Takeaway: Efficient cross-sectional slicing and slicing-by-category allow instantaneous portfolio metrics grouping.

Chapter 11: Time Series

Core Focus: Datetime handling, resampling, shifting (lead/lag), rolling-window calculations, and expanding-window operations.

Key Takeaway: Financial temporal alignment requires managing trading calendars, business days, and look-ahead bias prevention.

Chapter 12: Advanced pandas

Core Focus: Categorical types, advanced groupby usage, method chaining (.pipe, .assign), and performance optimizations.

Key Takeaway: Method chaining keeps transformation pipelines declarative, testable, and memory-efficient.
